```python
# Import packages
import plotly.express as px
import plotly.graph_objects as go
import pandas as pd

# Etiquetas y definición de colores
AUTOR ="Samuel Aguilar"
BG,INK,MUTE,GRIS,NAR = "#F5F3EF", "#1A1A1A", "#8A8680", "#DAD6CF", "#E8702A"

# Cargamos data set gapminder
df = px.data.gapminder().query("year == 2007")
# Imprimimos los países que tienen datos
print(len(df), "países con datos")
```

```python
# Pintamos la tierra base de NARANJA (=NULL) y lo que tiene datos en gris

fig = go.Figure(go.Choropleth(
                locations=df.iso_alpha, z=[1]*len(df),
                colorscale=[[0,GRIS],[1,GRIS]], showscale=False,
                marker_line_color=BG, marker_line_width=0.5
                ))
fig.update_geos(
    projection_type="natural earth", bgcolor=BG, showland= True, landcolor=NAR,
    showcountries=True, countrycolor=BG, showcoastlines=False, showframe=False,
    showocean = False, lataxis_range=[-58,85],
)
```

```python
# Realizamos el mapa con el layout
txt = lambda x,y,s,size, color, bold=False, anchor="left": fig.add_annotation(
    x=x, y=y, xref="paper",yref="paper",xanchor=anchor,
    showarrow=False, text=f"<b>{s}</b>" if bold else s,
    font=dict(size=size,color=color))

txt(0,1.22, "#30DAYMAPCHALLENGE | DÍA 18 : NULL", 16,  NAR, bold=True)
txt(1,1.22, AUTOR, 15, NAR, anchor="right")
txt(0,1.33, "NULL", 40, INK, bold=True)
txt(0,1.03, "Los países que no existen en el dataset Gapminder", 15, MUTE)
txt(0,-0.05, "<span style='color:%s'> Con datos </span>  <span style='color:%s'> NULL </span>" % (MUTE,NAR),12,GRIS)
txt(0,-0.06, "Datos: Gapminder (plotly.express) 2007", 10,MUTE,anchor="right")

# Layout configuration
fig.update_layout(paper_bgcolor=BG, width=1200, height=760,
                  margin=dict(l=40,r=40,t=190,b=60),
                  font=dict(family="Helvetica, Arial"))

# Map
fig.show()
```
